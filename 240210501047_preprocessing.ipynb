```python
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

stopword = StopWordRemoverFactory().create_stop_word_remover()
stemmer = StemmerFactory().create_stemmer()

def preprocess_text(text):
    text_clean = text.lower()
    text_clean = re.sub(r'[^a-z\s]', '', text_clean)
    
    tokens_before = text_clean.split()
    
    text_processed = stopword.remove(text_clean)
    text_processed = stemmer.stem(text_processed)
    
    tokens_after = text_processed.split()
    
    return text_processed, tokens_before, tokens_after

dokumen = [
    "Mahasiswa Teknik Komputer UNM sedang belajar Sistem Temu Kembali Informasi 101.",
    "Apakah jaringan komputer di kampus sudah menggunakan fiber optik yang super cepat?",
    "Teknologi AI saat ini berkembang sangat pesat di tahun 2024!",
    "Pengumuman: Libur semester ganjil akan dimulai pada tanggal 20 Desember mendatang.",
    "Kecerdasan buatan membantu memproses data besar secara efisien dan akurat."
]

for i, doc in enumerate(dokumen):
    hasil, t_before, t_after = preprocess_text(doc)
    
    print(f"--- Dokumen {i+1} ---")
    if i < 2:
        print(f"Original: {doc}")
        print(f"Preprocessed: {hasil}")
        
    jml_sebelum = len(t_before)
    jml_sesudah = len(t_after)
    persen_turun = ((jml_sebelum - jml_sesudah) / jml_sebelum) * 100 if jml_sebelum > 0 else 0
    
    print(f"Token sebelum: {jml_sebelum}")
    print(f"Token sesudah: {jml_sesudah}")
    print(f"Persentase pengurangan: {persen_turun:.2f}%\n")
```
